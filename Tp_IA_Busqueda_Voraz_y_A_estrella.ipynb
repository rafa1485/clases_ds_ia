Regla de desempate: ante prioridades iguales, se extrae el que se insertó antes (orden de inserción, FIFO). Deben respetarla todos los

Nota: un grafo dirigido representa una relación asimétrica: S→A permite ir de S a A, pero no de A a S. El problema formal es P=
(S,A,T,s0,G,c) con estados {S,A,B,C,D,G}, inicio S, objetivo {G}, transiciones las aristas listadas y costo c el valor de cada arista.

# RESOLUCION TP FLORENCIA SILVERO

```python
import heapq

# -------------------------------------------------------------------------
# 1. REPRESENTACIÓN DEL GRAFO Y HEURÍSTICA
# -------------------------------------------------------------------------
# Grafo como diccionario de adyacencia: nodo -> [(sucesor, costo), ...]
GRAFO = {
    'S': [('A', 2), ('B', 2)],
    'A': [('C', 2), ('D', 5)],
    'B': [('D', 2)],
    'C': [('G', 3)],
    'D': [('G', 6)],
    'G': []
}

HEURISTICA = {
    'S': 7,
    'A': 5,
    'B': 7,
    'C': 3,
    'D': 6,
    'G': 0
}

ESTADO_INICIAL = 'S'
ESTADO_OBJETIVO = 'G'

# -------------------------------------------------------------------------
# 2. FUNCIÓN DE RECONSTRUCCIÓN DE CAMINO
# -------------------------------------------------------------------------
def reconstruir_camino(nodo):
    camino = []
    actual = nodo
    while actual is not None:
        camino.append(actual['estado'])
        actual = actual['padre']
    return camino[::-1]

# -------------------------------------------------------------------------
# 3. IMPLEMENTACIÓN GENÉRICA DE BÚSQUEDA
# -------------------------------------------------------------------------
def busqueda_grafo(tipo_algoritmo):
    """
    Tipo algoritmo: 'UCS', 'VORAZ', 'A_ESTRELLA'
    """
    print("=" * 65)
    print(f" EJECUTANDO BÚSQUEDA: {tipo_algoritmo}")
    print("=" * 65)

    # Frontera: cola de prioridad implementada con heapq
    # Cada elemento almacenará: (prioridad, orden_insercion, nodo)
    frontera = []
    orden_insercion = 0

    # Diccionario de control mejor_g
    mejor_g = {}

    # Estadísticas
    estados_generados = 0
    estados_expandidos = 0
    frontera_maxima = 0
    reaperturas = 0

    # Crear nodo raíz
    g_raiz = 0
    h_raiz = HEURISTICA[ESTADO_INICIAL]
    raiz = {
        'estado': ESTADO_INICIAL,
        'padre': None,
        'accion': None,
        'g': g_raiz,
        'h': h_raiz
    }

    # Asignar prioridad según el algoritmo
    if tipo_algoritmo == 'UCS':
        prioridad_raiz = g_raiz
    elif tipo_algoritmo == 'VORAZ':
        prioridad_raiz = h_raiz
    elif tipo_algoritmo == 'A_ESTRELLA':
        prioridad_raiz = g_raiz + h_raiz

    heapq.heappush(frontera, (prioridad_raiz, orden_insercion, raiz))
    orden_insercion += 1
    estados_generados += 1
    mejor_g[ESTADO_INICIAL] = g_raiz

    paso = 0
    while frontera:
        # Registrar tamaño máximo de frontera
        if len(frontera) > frontera_maxima:
            frontera_maxima = len(frontera)

        # Imprimir contenido de la frontera para la traza
            traza_frontera = [f"({elem[2]['estado']}, p={elem[0]})" for elem in frontera]
            print(f"Paso {paso} | Frontera actual: {traza_frontera}")

        # Extraer el nodo con menor prioridad (desempate FIFO automático por orden_insercion)
        prio_actual, _, nodo_actual = heapq.heappop(frontera)
        estado_actual = nodo_actual['estado']

        # Descarte de entrada obsoleta si su g es peor que el mejor registrado
        if nodo_actual['g'] > mejor_g[estado_actual]:
            print(f"  -> [DESCARTE OBSOLETO] Estado {estado_actual} con g={nodo_actual['g']} (Mejor conocido: {mejor_g[estado_actual]})\n")
            continue

        print(f"  -> Extrayendo estado: {estado_actual} (g={nodo_actual['g']}, h={nodo_actual['h']}, prio={prio_actual})")

        # Prueba del objetivo AL EXTRAER
        if estado_actual == ESTADO_OBJETIVO:
            camino = reconstruir_camino(nodo_actual)
            costo = nodo_actual['g']
            print(f"\n¡OBJETIVO ALCANZADO!")
            print(f"Camino: {' -> '.join(camino)}")
            print(f"Costo Total: {costo}")
            print(f"Estados Generados: {estados_generados}")
            print(f"Estados Expandidos: {estados_expandidos}")
            print(f"Frontera Máxima: {frontera_maxima}")
            print(f"Reaperturas: {reaperturas}\n")

            return {
                'camino': ' -> '.join(camino),
                'costo': costo,
                'generados': estados_generados,
                'expandidos': estados_expandidos,
                'frontera_max': frontera_maxima,
                'reaperturas': reaperturas
            }

        estados_expandidos += 1

        # Relajar sucesores
        for sucesor, costo_arista in GRAFO[estado_actual]:
            nuevo_g = nodo_actual['g'] + costo_arista
            nuevo_h = HEURISTICA[sucesor]

            # Verificar si se encontró una mejora sobre mejor_g
            if sucesor not in mejor_g or nuevo_g < mejor_g[sucesor]:
                if sucesor in mejor_g:
                    # Si el estado ya había sido visitado previamente y se descubrió un camino mejor
                    reaperturas += 1

                mejor_g[sucesor] = nuevo_g

                nuevo_nodo = {
                    'estado': sucesor,
                    'padre': nodo_actual,
                    'accion': f"{estado_actual} -> {sucesor}",
                    'g': nuevo_g,
                    'h': nuevo_h
                }

                if tipo_algoritmo == 'UCS':
                    prio = nuevo_g
                elif tipo_algoritmo == 'VORAZ':
                    prio = nuevo_h
                elif tipo_algoritmo == 'A_ESTRELLA':
                    prio = nuevo_g + nuevo_h

                heapq.heappush(frontera, (prio, orden_insercion, nuevo_nodo))
                orden_insercion += 1
                estados_generados += 1

        paso += 1

    return None

# -------------------------------------------------------------------------
# 4. EJECUCIÓN Y TABLA COMPARATIVA
# -------------------------------------------------------------------------
if __name__ == "__main__":
    res_ucs = busqueda_grafo('UCS')
    res_voraz = busqueda_grafo('VORAZ')
    res_a = busqueda_grafo('A_ESTRELLA')

    print("=" * 65)
    print(" TABLA COMPARATIVA DE RESULTADOS")
    print("=" * 65)
    print(f"{'Métrica / Algoritmo':<25} | {'UCS':<12} | {'Voraz':<12} | {'A*':<12}")
    print("-" * 65)
    print(f"{'Camino':<25} | {res_ucs['camino']:<12} | {res_voraz['camino']:<12} | {res_a['camino']:<12}")
    print(f"{'Costo':<25} | {res_ucs['costo']:<12} | {res_voraz['costo']:<12} | {res_a['costo']:<12}")
    print(f"{'Prioridad':<25} | {'g':<12} | {'h':<12} | {'g+h':<12}")
    print(f"{'Expandidos antes de G':<25} | {res_ucs['expandidos']:<12} | {res_voraz['expandidos']:<12} | {res_a['expandidos']:<12}")
    print(f"{'Estados Generados':<25} | {res_ucs['generados']:<12} | {res_voraz['generados']:<12} | {res_a['generados']:<12}")
    print(f"{'Frontera Máxima':<25} | {res_ucs['frontera_max']:<12} | {res_voraz['frontera_max']:<12} | {res_a['frontera_max']:<12}")
    print(f"{'Reaperturas':<25} | {res_ucs['reaperturas']:<12} | {res_voraz['reaperturas']:<12} | {res_a['reaperturas']:<12}")
    print("=" * 65)
```

### Respuestas a las Preguntas de Análisis

1. ¿Por qué voraz y A* coinciden en este grafo? ¿Qué condición del grafo y de la heurística lo explica?

Coinciden porque la heurística $h(n)$ proporcionada está perfectamente orientada para guiar la búsqueda directamente por el camino óptimo hacia $G$. Además, los costos reales de los arcos en la rama superior ($S \to A \to C \to G$, costo total $7$) coinciden exactamente con la dirección en la que $h(n)$ decrece más rápido. La heurística no comete "errores" ni engaña al algoritmo voraz haciéndolo elegir una rama más larga o costosa.

2. ¿Garantiza voraz devolver el camino de menor costo en general? Justificá con la propiedad de su prioridad.

No, no lo garantiza. La búsqueda voraz utiliza exclusivamente $h(n)$ como función de prioridad. Su único criterio es tomar el nodo que parece estar más cerca del objetivo, ignorando completamente el costo acumulado recorrido hasta el momento ($g(n)$). Si una rama ofrece estados con heurísticas aparentemente bajas pero sus transiciones tienen costos sumamente elevados, Voraz elegirá esa rama ciega y terminará devolviendo una solución subóptima

3. ¿Qué ocurre si se usa $h(n) = 0$ en A*? ¿Con qué algoritmo coincide entonces?

Si $h(n) = 0$ para todos los estados, la función de prioridad de A* pasa a ser:$$f(n) = g(n) + h(n) = g(n) + 0 = g(n)$$En este caso, la prioridad depende únicamente del costo acumulado $g(n)$, por lo que A* se vuelve conceptual e implementativamente idéntico a Costo Uniforme (UCS).

4. ¿Hubo reaperturas en UCS? ¿Y en A* y voraz? ¿Por qué?

No hubo reaperturas en ninguno de los tres algoritmos.En UCS, la primera vez que un nodo se extrae o procesa bajo un grafo con costos no negativos, se asegura de haber hallado la distancia más corta posible.En A*, la heurística dada es consistente/admisible (para toda transición $u \to v$, se cumple $h(u) \le c(u, v) + h(v)$). Por ejemplo, en $A \to C$, $h(A)=5 \le 2 + h(C)=2+3=5$. Como la heurística no sobreestima ni cae en inconsistencias, A* no necesita reabrir ningún nodo cerrado.En Voraz, la estructura simple del grafo y el orden de expansión impidieron que se generara un camino secundario más barato a un estado ya visitado.

5. ¿"Expandir menos estados" significa "camino más barato"? Relacionalo con lo que muestran UCS y voraz aquí.

No, expandir menos estados no implica encontrar el camino más barato.La cantidad de estados expandidos refleja la eficiencia o rapidez en la búsqueda.El costo del camino refleja la admisibilidad u optimalidad del algoritmo.En este grafo en particular, Voraz expande $3$ estados y halla el camino de costo $7$, mientras que UCS expande $4$ estados para hallar el mismo costo $7$. Sin embargo, esto es circunstancial. En el caso general, la Búsqueda Voraz suele expandir muchísimos menos estados que UCS porque va directo al objetivo, pero frecuentemente devuelve caminos subóptimos y más caros, mientras que UCS garantiza siempre la solución óptima a costa de explorar más partes del grafo.
